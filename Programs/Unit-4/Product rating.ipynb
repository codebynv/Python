```python
with open("product.txt","w") as f:
    f.write("A,50\n")
    f.write("B,90\n")
    f.write("C,70\n")
    f.write("D,30\n")
```

```python
with open("product.txt","r") as f:
    content=f.read()
    print(content)
```

```python
import matplotlib.pyplot as plt

products=[]
votes=[]

with open("product.txt","r") as f:
    for line in f:
        p,v=line.split(",")
        products.append(p)
        votes.append(v)

myexplode=[]

mvote=max(votes)
for i in votes:
    if i==mvote:
        myexplode.append(0.1)
    else:
         myexplode.append(0)
            
plt.pie(votes, labels=products,explode=myexplode,autopct="%1.1f%%")
plt.title("Product Popularity")
plt.show()
```
