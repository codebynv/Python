```python
import numpy as np

a=np.arange(10,50)
print(a)
```

```python
a=np.array([5,10,15,20,25])
print(a[2])
```

```python
a=np.array([[5,10,15,20,25],[55,110,151,201,25],[65,80,15,20,25]])
print(a)
print("After Slicing")
print(a[:2])
```

```python
a=np.array([5,10,15,20,25,30])
b=a.reshape(2,3)
print(b)
```

```python
a=np.array([5,10,15,20,25,30])
b=np.array([55,510,515,520,525,530])
c=np.concatenate((a,b))
print(c)
```

```python
a=np.array([5,10,15,20,25,30,70])
i=np.where(a>25)
print(i)
```

```python
b=np.array([55,510,515,520,525,530])
c=np.sort(b)
print(c)
```

```python
a=np.array([12, 15, 20, 25, 30, 35, 40])
print(a)

new=[]
for i in a:
    if i>25:
        new.append(i)
        
new_array=np.array(new)
print(new_array)
print(type(new_array))
```

```python
a=np.array([5,10,15,20])
b=np.array([55,510,515,520])
c=a/b
print(c)
```

```python
a=np.array([[5,10],[55,110],[65,80]])
print(a)

print("After Transpose")
print(a.T)
```

```python
a=np.array([55,510,515,520])
print(np.min(a))
print(np.max(a))
print(np.mean(a))
```

```python
a=np.array([[5,10],[6,11]])
print(a)
b=np.array([[55,101],[65,111]])
print(b)
print("Aftre Concatenation")
c=np.concatenate((a,b))
print(c)
print("Aftre horizontal Concatenation")
c=np.concatenate((a,b),axis=1)
print(c)
```

```python
a=np.array([25000, 32000, 28000, 45000, 30000, 27000])
print(a)
increment=np.where(a<30000,a*0.1,a*0.05)
print("increment as per policy",increment)
```
