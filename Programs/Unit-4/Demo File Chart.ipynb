```python
with open("product_votes.txt", "w") as file:
    file.write("ProductA,50\n")
    file.write("ProductB,30\n")
    file.write("ProductC,20\n")
    file.write("ProductD,40\n")
```

```python
import matplotlib.pyplot as plt

products = []
votes = []


with open("product_votes.txt", "r") as file:
    for line in file:
        name, vote = line.strip().split(",")
        products.append(name)
        votes.append(int(vote))


max_votes = max(votes)
explode = [0.1 if v == max_votes else 0 for v in votes]


plt.pie(votes, labels=products, explode=explode,
        autopct='%1.1f%%', startangle=90)

plt.title("Product Popularity Poll")

plt.show()
```
