```python
mydb=mysql.connector.connect(host="localhost",user="root",password="",database="db1")
mycursor=mydb.cursor()
mycursor.execute("create table student(name varchar(20), password varchar(20))")
```

```python
from tkinter import *

root = Tk()
l1 =Label(root,text="Name")
l1.grid(row=0,column=0)

l1 =Label(root,text="password")
l1.grid(row=1,column=0)

e1=Entry(root)
e1.grid(row=0,column=1)

e2=Entry(root)
e2.grid(row=1,column=1)

def insert():
    name=e1.get()
    password=e2.get()
    sql="insert into student(name,password) values (%s,%s)"
    val=(name,password)
    mycursor.execute(sql,val)
    mydb.commit()

def select():
    mycursor.execute("select * from student")
    result=mycursor.fetchall()
    for i in result:
        print(i)
    
def delete():
    uname=input("enter the name")
    sql="delete from student where name=%s"
    val=(uname,)
    mycursor.execute(sql,val)
    mydb.commit()
    
def update():
    uname=input("enter the name for which you want to update password")
    new=input("enter new password")
    sql="update student set password=%s where name=%s"
    val=(new,uname)
    mycursor.execute(sql,val)
    mydb.commit()

    
btn=Button(root,text="submit",command=insert)
btn.grid(row=3,column=1)

btn=Button(root,text="show",command=select)
btn.grid(row=4,column=1)


btn=Button(root,text="delete",command=delete)
btn.grid(row=5,column=1)

btn=Button(root,text="update",command=update)
btn.grid(row=6,column=1)
root.mainloop()
```
